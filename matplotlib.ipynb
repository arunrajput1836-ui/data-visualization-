```python
# subplot
x = [1,2,3,4,5]
sales_a =[10,20,15,25,30]
sales_b=[5,15,10,20,25]

plt.subplot(1,2,1)
plt.plot(x,sales_a,marker='o')
plt.title("sales a")

plt.subplot(1,2,2)
plt.bar(x,sales_b,color ='red')
plt.show()
```

```python
 # Area chart
 import matplotlib.pyplot as plt
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
sales = [100, 150, 200, 180, 220, 300]
sales2 = [200, 250, 300, 280, 320, 400]

# Simple area chart

plt.plot(months, sales, color="blue")
plt.plot(months, sales2, color="red")
plt.fill_between(months, sales2, color="green", alpha=0.5)
plt.fill_between(months, sales, color="blue", alpha=0.5)
plt.title("Simple Area Chart - Monthly Sales")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.show()
```

```python
# bubble chart
x = [5, 7, 8, 7, 6, 9, 5, 8]
y = [99, 86, 87, 88, 100, 86, 103, 87]
sizes = [20, 50, 100, 200, 500, 1000, 60, 90]  # Bubble sizes

# Bubble chart
plt.scatter(x, y, s=sizes, alpha=0.5,color ="red",edgecolors="black",linewidth = 5)
plt.title("Simple Bubble Chart")
plt.xlabel("X Values")
plt.ylabel("Y Values")
plt.show()
```

```python
# bubble chart customized
x = [5, 7, 8, 7, 6, 9, 5, 8]
y = [99, 86, 87, 88, 100, 86, 103, 87]
sizes = [200, 400, 600, 800, 1000, 1200, 500, 700]
colors = ["red", "blue", "green", "purple", "orange", "brown", "pink", "cyan"]

plt.figure(figsize=(7,5))

# Customized bubble chart
plt.scatter(x, y, s=sizes, color=colors, alpha=0.6, edgecolors="black", linewidth=5)

plt.title("Customized Bubble Chart", fontsize=14, fontweight="bold")
plt.xlabel("X Axis", fontsize=14, fontweight="bold")
plt.ylabel("Y Axis", fontsize=14, fontweight="bold")
plt.grid(linestyle="--", alpha=0.7)

plt.show()
```
