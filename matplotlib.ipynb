```python
import matplotlib.pyplot as plt

x = [2020, 2022, 2024, 2026]
y = [10, 20, 0, 4 ]


plt.plot(x, y)
plt.show()
```

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4]
y = [10, 20, 15, 30]

plt.plot(
    x, y,
    marker="o",
    linestyle = "-.",
    markersize=20,
    linewidth=20
)

plt.show()
```

```python
import matplotlib.pyplot as plt
x=[10, 20, 30 ,40 ,50]
y=[5, 50, 1, 20, 2]
plt.plot(x,y, marker = "*", markersize = 5,linestyle = "--", color = "green",linewidth =2 )

plt.show()
```

```python
import matplotlib.pyplot as plt
subjects = [ "maths", "english", "hindi", "science"]
marks = [70, 85, 90, 75]
plt.bar(subjects,marks,color ="green",width =0.5,edgecolor="red",alpha=0.8)
plt.title("my marks")
plt.xlabel(subjects)
plt.ylabel(marks)
plt.show
```

```python
import matplotlib.pyplot as plt
subjects = [ "java","python","ml","maths"]
marks = [60,75,90,85]
plt.scatter(subjects,marks,color="red",alpha =0.8,marker="*",s=100)
plt.title("my marks")
plt.xlabel(subjects)
plt.ylabel(marks)
plt.show()
```

```python
import matplotlib.pyplot as plt
marks = [45,50,60,55,70,62,72,80,75,85,90]
plt.hist(marks,color = "red",bins = 10,edgecolor ="black")
plt.title("my marks")
plt.xlabel("marks")
plt.show()
```
